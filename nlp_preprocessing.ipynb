```python
import pandas as pd
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

# Download necessary NLTK datasets (Updated for the newest NLTK versions)
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng') # Added this line for POS Tagging
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
print("--- DEMONSTRATING NLP TECHNIQUES ---")
sample_text = "How do I reset my billing passwords? The shipping details are arriving late!"

# 1. Tokenization
tokens = word_tokenize(sample_text)
print("1. Tokens:\n", tokens, "\n")

# 2. Stopword Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words and word.isalnum()]
print("2. No Stopwords:\n", filtered_tokens, "\n")

# 3. POS Tagging
pos_tags = nltk.pos_tag(filtered_tokens)
print("3. POS Tags:\n", pos_tags, "\n")

# 4. Stemming
stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]
print("4. Stemmed:\n", stemmed_words, "\n")

# 5. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_tokens]
print("5. Lemmatized:\n", lemmatized_words, "\n")
```

```python
print("--- EVALUATING ON DATASET ---")
data = {
    'text': [
        "How do I reset my password? I can't log in.",
        "My screen is frozen and the software keeps crashing.",
        "Where is my invoice for last month?",
        "I was double-charged on my credit card, please refund.",
        "My package was supposed to arrive yesterday.",
        "Can I change the delivery address for my order?"
    ],
    'category': [
        "technical support", 
        "technical support", 
        "billing", 
        "billing", 
        "shipping", 
        "shipping"
    ]
}

df = pd.DataFrame(data)
```

```python
def preprocess_text_for_ml(text):
    """Cleans text for the machine learning models in Phase 2."""
    tokens = word_tokenize(text.lower())
    filtered_tokens = [word for word in tokens if word not in stop_words and word.isalnum()]
    # We use Lemmatization over Stemming for better accuracy in Phase 2
    lemmatized_tokens = [lemmatizer.lemmatize(word) for word in filtered_tokens]
    return " ".join(lemmatized_tokens)

# Apply the preprocessing
df['cleaned_text'] = df['text'].apply(preprocess_text_for_ml)

# Display the final dataset ready for Phase 2
print("Dataset Ready for Feature Extraction (TF-IDF):")
print(df[['category', 'cleaned_text']])
```
