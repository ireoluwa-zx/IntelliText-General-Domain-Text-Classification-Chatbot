```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# 1. The Dataset (Same as Phase 2)
data = {
    'cleaned_text': [
        "reset password", "forgot password", "software crash", "software update fail", "cannot access account", "account password reset",
        "invoice last month", "missing invoice", "double charge credit card", "refund credit card", "billing cycle charge", "refund invoice",
        "package arrive late", "track package", "change delivery address", "track delivery status", "shipping delay", "order arrive late"
    ],
    'category': [
        "technical support", "technical support", "technical support", "technical support", "technical support", "technical support",
        "billing", "billing", "billing", "billing", "billing", "billing",
        "shipping", "shipping", "shipping", "shipping", "shipping", "shipping"
    ]
}

df = pd.DataFrame(data)

# 2. Vectorization & Split
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['cleaned_text'])
y = df['category']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# ==========================================
# 3. TUNING NAIVE BAYES
# Parameter to tune: 'alpha' (Laplace smoothing). 
# This helps the model handle words in the test set it didn't see in the training set.
# ==========================================
print("--- NAIVE BAYES TUNING ---")
nb_param_grid = {'alpha': [0.1, 0.5, 1.0, 2.0, 5.0, 10.0]}

# GridSearchCV will test every 'alpha' value using cross-validation
nb_grid = GridSearchCV(MultinomialNB(), nb_param_grid, cv=3)
nb_grid.fit(X_train, y_train)

best_nb = nb_grid.best_estimator_
nb_tuned_predictions = best_nb.predict(X_test)

print(f"Best Alpha Setting: {nb_grid.best_params_['alpha']}")
print(f"Tuned Accuracy: {accuracy_score(y_test, nb_tuned_predictions) * 100:.2f}%\n")


# ==========================================
# 4. TUNING DECISION TREE
# Parameters to tune: 
# 'criterion' (Entropy vs Gini), 
# 'max_depth' (How many questions the tree can ask before stopping to prevent overfitting)
# ==========================================
print("--- DECISION TREE TUNING ---")
dt_param_grid = {
    'criterion': ['gini', 'entropy'],
    'max_depth': [None, 2, 5, 10],
    'min_samples_split': [2, 5]
}

dt_grid = GridSearchCV(DecisionTreeClassifier(random_state=42), dt_param_grid, cv=3)
dt_grid.fit(X_train, y_train)

best_dt = dt_grid.best_estimator_
dt_tuned_predictions = best_dt.predict(X_test)

print(f"Best Settings: {dt_grid.best_params_}")
print(f"Tuned Accuracy: {accuracy_score(y_test, dt_tuned_predictions) * 100:.2f}%")
```
