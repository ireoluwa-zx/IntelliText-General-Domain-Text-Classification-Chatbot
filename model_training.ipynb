```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# 1. Improved Dataset (with overlapping keywords so the models can learn patterns)
data = {
    'cleaned_text': [
        "reset password", "forgot password", "software crash", "software update fail", "cannot access account", "account password reset",
        "invoice last month", "missing invoice", "double charge credit card", "refund credit card", "billing cycle charge", "refund invoice",
        "package arrive late", "track package", "change delivery address", "track delivery status", "shipping delay", "order arrive late"
    ],
    'category': [
        "technical support", "technical support", "technical support", "technical support", "technical support", "technical support",
        "billing", "billing", "billing", "billing", "billing", "billing",
        "shipping", "shipping", "shipping", "shipping", "shipping", "shipping"
    ]
}

df = pd.DataFrame(data)

# 2. Feature Extraction (TF-IDF)
vectorizer = TfidfVectorizer()
X = vectorizer.fit_transform(df['cleaned_text'])
y = df['category']

# 3. Train-Test Split (Testing on roughly 4 items this time)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 4. Train and Evaluate Naive Bayes
nb_classifier = MultinomialNB()
nb_classifier.fit(X_train, y_train)
nb_predictions = nb_classifier.predict(X_test)

print("--- NAIVE BAYES CLASSIFIER ---")
print(f"Accuracy: {accuracy_score(y_test, nb_predictions) * 100:.2f}%\n")
print(classification_report(y_test, nb_predictions, zero_division=0))

# 5. Train and Evaluate Decision Tree
dt_classifier = DecisionTreeClassifier(random_state=42)
dt_classifier.fit(X_train, y_train)
dt_predictions = dt_classifier.predict(X_test)

print("\n--- DECISION TREE CLASSIFIER ---")
print(f"Accuracy: {accuracy_score(y_test, dt_predictions) * 100:.2f}%\n")
print(classification_report(y_test, dt_predictions, zero_division=0))
```
